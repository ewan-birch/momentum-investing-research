# Portfolio Creation

## How are Momentum Portfolios Created

When we create Momentum portfolios we first have to measure the momentum of the stocks we are selecting between. Momentum tends to be measured in 3 time intervals: 3-month, 6-month and 12-month momentums, note that we exclude the previous month in our considerations as it can play too large a role in price differentials and by looking at larger time periods we hope to 'see' the fundermental momentum which underlies the holdings growth in value.

## Momentum Scoring and Choice of Stock

When picking the stocks with the most momentum it is often the case that we consider a weighted aggregate of the 3,6 and 12 month momentums as this in theory gives us the stocks with the best fundamental momentum, whilst making sure that we choose stocks with high current momentum.

- Aggregate Score = (0.5 * 3_month) + (0.3 * 6_month) + (0.2 * 12_month)

So we have weighted the score so that the more recent momentum has a greater effect on the portfolio, whilst making sure that its not just a 'random' change in value that leads to a stock being added to our portfolio.

## Portfolio Construction Theory

Once we have found the aggregate scores of each of the stocks each month, we will rank them and then chose the best 25. We do this so that we are choosing the stocks with the greatest momentum, but we choose a variety of stocks because diversification if crutial in a portfolio as it greatly reduces risk as the volatility of a single stock is much greater than that of a collection of stocks.

We then normalise the weights of these stocks in the portfolio by calculating the total aggregate momentum score of all the stocks in our portfolio for a given month and then calculate the percentage of this total score that each of the stocks make up. By constructing the portfolio this way we are once again rewarding greater momentum and assuming that the greater the momentum of a stock the greater the subsequent returns it shall grant us.
