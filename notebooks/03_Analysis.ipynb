# Historical Returns of the Momentum Portfolio

Having constructed the momentum portfolios and obtained the historical price data for the constituent stocks, we calculate the returns generated by the strategy over the period from 30 January 2009 to 30 October 2024.

As the portfolio is rebalanced each month, we calculate portfolio turnover to measure the extent of trading required between successive rebalancing periods. We use one-way portfolio turnover, calculated by summing the absolute changes in portfolio weights between two consecutive rebalancing periods and dividing by 2. The division by 2 converts the sum of absolute weight changes, which captures both purchases and sales, into one-way turnover representing the proportion of the portfolio that is reallocated.

The gross portfolio return for each period is calculated using the returns of the individual stocks and their corresponding portfolio weights.

Monthly gross returns are subsequently compounded over the full backtest period to calculate the cumulative return of the momentum strategy.

```python
# Imported libraries

import sys
from pathlib import Path
import pickle 
import pandas as pd
```

```python
base_dir = Path.cwd().parent
data_dir = base_dir / "data"
sys.path.append(str(base_dir))

with open(data_dir / "processed" / "prices.pkl", "rb") as f:
    prices = pickle.load(f)

from src.backtest import(
    monthly_portfolio_return,
    periodic_momentum_returns
)

from src.portfolio import normalised_weightings

gross_return, turnover, new_weights = monthly_portfolio_return(
        "2023-06-20",
        "2023-07-20",
        normalised_weightings("2023-06-20", prices)["Weights"]
    )

print("For the Month: 2023-06-20 - 2023-07-20 ")
print(f"Gross Returns: {gross_return}")
print(f"Turnover: {turnover}")
print(f"Net Weights: \n {new_weights}")

total_gross_return, total_net_return, total_turnover = periodic_momentum_returns("2020-03-28", "2024-06-28", 0.001)

print("\nFor the period: 2020-03-28 - 2024-06-28")
print(f"Total Gross Return: {total_gross_return}")
print(f"Total Net Return: {total_net_return}")
print(f"Total Turnover: {total_turnover}")
```
